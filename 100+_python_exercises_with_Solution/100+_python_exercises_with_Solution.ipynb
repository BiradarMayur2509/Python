## Exercise 2: Swap Two Variables Without a Third Variable

**Objective**: Swap two values without using a temporary variable.

Task:
• Swap the values of students_morning = 15 and students_evening = 25 without us
ing an extra variable.
    
    Expected Output:
        Before Swap: Morning Batch = 15, Evening Batch = 25
        After Swap: Morning Batch = 25, Evening Batch = 15


```python
morning_batch = 15
evening_batch = 25

print(f"Before Swap : Morning Batch : {morning_batch}, Evening Batch = {evening_batch}")

morning_batch, evening_batch = evening_batch, morning_batch

print(f"After Swap : Morning Batch = {morning_batch}, Evening Batch = {evening_batch}")


```
