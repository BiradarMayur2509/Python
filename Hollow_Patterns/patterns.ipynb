## EX1: Python Program to Print Hollow Square Pattern Using 

#### Solution:

```python
def hollow_square(n):
    
    for i in range(1,n+1):
        for j in range(1,n+1):
            if i == 1 or i == n or j == 1 or j == n:
                print("*", end= " ")
            else:
                print(" ", end= " ")
        print()
        
size = 7
hollow_square(size)
        
```

# EX2 : Python Program to Print Hollow Square Pattern Using String Multiplication

#### Solution:


```python
def hollow_square_2(n):
    
    print("* " * n)
    
    for i in range(n-2):
        print("* " + "  " * (n-2) + "*")
        
    if n > 1:
        print("* " * n)
        
size = 5
hollow_square_2(size)
```
