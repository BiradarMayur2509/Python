## Q6: Employee Management System (Inheritance & Method Overriding)

### Problem Statement

Create a base class `Employee` with:

- Attributes:
  - `name`
  - `salary`

- Method:
  - `calculate_salary()`

---

### Child Classes

1. **Manager**
   - Additional Attribute:
     - `bonus`
   - Override `calculate_salary()` to include bonus.

2. **Developer**
   - Additional Attribute:
     - `programming_language`
   - Customize `calculate_salary()` if needed.

---

### 🎯 Tasks

#### ✅ Task 1: Create Base Class
- Define class `Employee`
- Add constructor to initialize:
  - `name`
  - `salary`
- Implement method:
  - `calculate_salary()` → returns base salary

---

#### ✅ Task 2: Implement Manager Class
- Inherit from `Employee`
- Add attribute:
  - `bonus`
- Override method:
  - `calculate_salary()`
    - Return: `salary + bonus`

---

#### ✅ Task 3: Implement Developer Class
- Inherit from `Employee`
- Add attribute:
  - `programming_language`
- Use base class `calculate_salary()` OR extend if needed

---

#### ✅ Task 4: Create Objects & Test
- Create:
  - One `Manager` object
  - One `Developer` object
- Call:
  - `calculate_salary()` for both
- Print results

---

#### ✅ Task 5 (Optional - Advanced)
- Add validation:
  - Salary should not be negative
  - Bonus should not be negative
- Add `__str__()` method for better output display

---

### Objective

- Demonstrate **inheritance**
- Apply **method overriding**
- Understand **polymorphism**
- Build a **real-world employee system**

```python
class Employee:
    def __init__(self,name, salary):
        if salary < 0:
            raise ValueError("Salary cannot be negative")
        self.name = name
        self.salary = salary
    
    def calculate_salary(self):
        return self.salary
    
    def __str__(self):
        return f"Employee: {self.name}, Salary: {self.salary}"
        
        
class Manager(Employee):
    def __init__(self,name,salary,bonus):
        super().__init__(name,salary)
        if bonus < 0:
            raise ValueError("Bonus cannot be negative")
        self.bonus = bonus
    
    #overridden methon     
    def calculate_salary(self):
        return self.salary + self.bonus
    
    def __str__(self):
        return f"Manager: {self.name}, Salary:{self.salary}, Bonus: {self.bonus}"
        
        
class Developer(Employee):
    def __init__(self,name, salary, programming_lang):
        super().__init__(name,salary)
        self.programming_lang = programming_lang
    
    def __str__(self):
        return (
            f"Developer: {self.name}, salary: {self.salary},"
            f"Language : {self.programming_lang}"
        )
        
        
# Testing 

manager = Manager("Ajay", 80000, 20000)
developer = Developer("Mayur", 60000, "Python")

employees = [manager, developer]

for emp in employees:
    print(emp)
    print(f"Calculated Salary: {emp.calculate_salary()}")
    print("-" * 40)
        
```
