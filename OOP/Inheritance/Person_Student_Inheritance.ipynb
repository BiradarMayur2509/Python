# Q2: Person → Student

Create a base class `Person` with:

- Attributes:
  - `name`
  - `age`

Create a child class:

- `Student` → inherits from `Person` and adds:
  - `grade`

### Method:
- `display_info()` → prints all details (name, age, grade)

### Task:
1. Define the `Person` class with `name` and `age`.
2. Create the `Student` class that inherits from `Person` and adds `grade`.
3. Implement the `display_info()` method in `Student`.
4. Create a `Student` object and display its information.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# child class : Student
class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)  # call parent constructor 
        self.grade = grade
    
    def display_info(self):
        print("Student Details:")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")
        
student1 = Student("Mayur", 26, "A")
student1.display_info()
```
