# Q1: Animal Hierarchy

Create a base class `Animal` with:

- Attribute:
  - `name`
- Method:
  - `speak()`

Create child classes:

- `Dog` → overrides `speak()` to print `"Bark"`
- `Cat` → overrides `speak()` to print `"Meow"`

### Task:
1. Define the `Animal` class.
2. Create `Dog` and `Cat` classes that inherit from `Animal`.
3. Override the `speak()` method in each child class.
4. Create objects of `Dog` and `Cat` and call their `speak()` methods.

```python
# Base class
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print("Animal makes a sound")

# child class: dog
class Dog(Animal):
    def speak(self):
        print("Bark")

#child class : cat 
class Cat(Animal):
    def speak(self):
        print("Meow")


dog = Dog("Buddy")
cat = Cat("Whiskers")

dog.speak()
cat.speak()   
```
