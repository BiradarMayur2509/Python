# Q3: Vehicle System

Create a base class `Vehicle` with:

- Attributes:
  - `brand`
  - `speed`

Create child classes:

- `Car` → inherits from `Vehicle` and adds:
  - `fuel_type`
- `Bike` → inherits from `Vehicle` and adds:
  - `type` (sports/normal)

### Method:
- `show_details()` → prints all details

### Task:
1. Define the `Vehicle` class.
2. Create `Car` and `Bike` classes that inherit from `Vehicle`.
3. Add specific attributes (`fuel_type`, `type`) in child classes.
4. Implement `show_details()` method.
5. Create objects and display their details.

```python
# Base class : Vehicle 
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
        
    def show_details(self):
        print("Show Details:")
        print(f"Brand: {self.brand}")
        print(f"Speed: {self.speed}")
        
# child class : Car
class Car(Vehicle):
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand, speed)
        self.fuel_type = fuel_type
    
    def show_details(self):
        super().show_details()
        print(f"Fuel Type: {self.fuel_type}")
        
# child class : Bike
class Bike(Vehicle):
    def __init__(self, brand, speed, bike_type):
        super().__init__(brand, speed)
        self.bike_type = bike_type
    
    def show_details(self):
        super().show_details()
        print(f"Type: {self.bike_type}")
        

# Example usage
car1 = Car("Toyota", 160, "Diesel")
bike1 = Bike("Honda", 120, "Sports")

car1.show_details()
print("________________")
bike1.show_details()
```
