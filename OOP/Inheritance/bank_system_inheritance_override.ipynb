## Q5: Bank System (Inheritance & Method Overriding)

### Problem Statement

Create a base class `BankAccount` with:

- Attribute:
  - `balance`

- Methods:
  - `deposit(amount)`
  - `withdraw(amount)`

---

### Child Classes

1. **SavingsAccount**
   - Additional Attribute:
     - `interest_rate`
   - Customize withdrawal behavior if needed.

2. **CurrentAccount**
   - Additional Attribute:
     - `overdraft_limit`
   - Override `withdraw()` to allow withdrawal beyond balance up to overdraft limit.

---

### 🎯 Tasks

1. Create a base class `BankAccount`:
   - Initialize `balance` using constructor.
   - Implement `deposit(amount)`:
     - Add amount to balance.
     - Validate that amount is positive.
   - Implement `withdraw(amount)`:
     - Allow withdrawal only if balance is sufficient.
     - Handle invalid inputs.

2. Create a class `SavingsAccount`:
   - Inherit from `BankAccount`.
   - Add `interest_rate` attribute.
   - (Optional) Override `withdraw()` if you want to restrict withdrawals.

3. Create a class `CurrentAccount`:
   - Inherit from `BankAccount`.
   - Add `overdraft_limit` attribute.
   - Override `withdraw(amount)`:
     - Allow withdrawal if:
       ```
       amount <= balance + overdraft_limit
       ```

4. Create objects of each class and test:
   - Deposit money
   - Withdraw money
   - Test overdraft behavior in `CurrentAccount`

---

### Objective

- Demonstrate **inheritance**
- Implement **method overriding**
- Apply **real-world banking logic**

````python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    
    def deposite(self,amount):
        if amount <= 0:
            print("Deposite amount must be positive")
            return
        self.balance += amount
        print(f"Deposited: {amount}, New Balance: {self.balance}")
        
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
            return
        if amount > self.balance:
            print("Insufficient balance.")
            return
        self.balance -= amount
        print(f"Withdrawn: {amount}, Remaining Balance: {self.balance}")
        

class SavingAccount(BankAccount):
    def __init__(self, balance=0, interest_rate=0.0):
        super().__init__(balance)
        self.interest_rate = interest_rate
    

class CurrentAccount(BankAccount):
    def __init__(self, balance=0, overdraft_limit=0):
        super().__init__(balance)
        self.overdraft_limit = overdraft_limit
        
    #overridden method
    def withdraw(self,amount):
        if amount <=0:
            print("Withdrawal amount must be positive")
            return
        if amount > (self.balance + self.overdraft_limit):
            print("Exceeded overfraft limit.")
            return
        self.balance -= amount
        print(f"Withdrawn: {amount}, Remaining Balance: {self.balance}")
        
# Testing the system
print("---- Savings Account ----")
sa = SavingAccount(1000, 5.0)
sa.deposite(500)
sa.withdraw(300)
sa.withdraw(1500) 

print("\n---- Current Account ----")
ca = CurrentAccount(1000, 500)
ca.deposite(200)
ca.withdraw(1300) #allowed (uses overdraft)
ca.withdraw(500)
        
````
