# Q7: Online Course Platform (Inheritance & Real-World OOP)

## Problem Statement

Design and develop an **Online Course Platform** using Object-Oriented Programming (OOP) concepts in Python.

Create a base class `User` with:

- Attributes:
  - `username`
  - `email`

---

## Child Classes

### 1. Student
- Additional Attribute:
  - `enrolled_courses`

- Method:
  - `enroll_course()`
    - Allows students to enroll in courses

---

### 2. Instructor
- Additional Attribute:
  - `created_courses`

- Method:
  - `create_course()`
    - Allows instructors to create new courses

---

## 🎯 Tasks

### ✅ Task 1: Create Base Class
- Define class `User`
- Add constructor to initialize:
  - `username`
  - `email`

---

### ✅ Task 2: Implement Student Class
- Inherit from `User`
- Add attribute:
  - `enrolled_courses`
- Implement method:
  - `enroll_course(course_name)`
    - Store enrolled course names

---

### ✅ Task 3: Implement Instructor Class
- Inherit from `User`
- Add attribute:
  - `created_courses`
- Implement method:
  - `create_course(course_name)`
    - Store created course names

---

### ✅ Task 4: Create Objects & Test
- Create:
  - One `Instructor` object
  - One `Student` object

- Perform:
  - Course creation by instructor
  - Course enrollment by student

- Display:
  - Created courses
  - Enrolled courses

---

### ✅ Task 5 (Optional - Advanced)

## Objective

- Demonstrate **inheritance in Python**
- Build reusable and scalable class structures
- Understand **real-world OOP implementation**
- Manage relationships between students and instructors
- Simulate a simple Learning Management System (LMS)
- Improve problem-solving and backend design concepts

```python
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email
    
class Student(User):
    def __init__(self, username,email):
        super().__init__(username, email)
        self.enrolled_courses = []
        
    def enroll_course(self, course_name):
        self.enrolled_courses.append(course_name)
        print(f"{self.username} enrolled in '{course_name}'")
        
class Instructor(User):
    def __init__(self,username, email):
        super().__init__(username, email)
        self.created_courses = []
        
    def create_course(self, course_name):
        self.created_courses.append(course_name)
        print(f"{self.username} created course '{course_name}'")
        
        
instructor1 = Instructor("Rahul", "rahul@edu.com")

# Instructor creates course
instructor1.create_course("Python Mastery")
instructor1.create_course("Data science Bootcamp")

print("-" * 40 )

# create student object 

student1 = Student("Mayur", "mayur@gmail.com")

# Student enrolls in courses
student1.enroll_course("Data science Bootcamp")
print("-" * 40 )


print("Student Enrolled Courses:")
print(student1.enrolled_courses)

```
