## Create a python program using a class named lapto.

    taks :
        - define a class Laptop
        - Inside it, create a   method     details (self,    brand, ram)
        - create teo different laptop object.
        - Call the details() method for each object, passing different values (ex: "Dell", "16GB" and "Hp", "8GB")

```python
class Laptop:
    
    def __init__(self, brand, ram, price, processor):
        self.brand = brand
        self.ram = ram
        self.price = price
        self.processor = processor
    
    def details(self):
        return f"""
    Brand: {self.brand}
    RAM: {self.ram} 
    Price: {self.price}  
    Processor: {self.processor} 
    ---------------------------"""

# Create objects
#l1 = Laptop("Dell", "16GB", "Rs70,000", "Intel i5")
# l2 = Laptop("HP", "8GB", "Rs50,000", "Ryzen5")

Laptops = [
    Laptop("Dell", "16GB", 70000, "Intel i5"),
    Laptop("HP", "8GB", 50000, "Ryzen 5"),
    Laptop("Lenovo", "12GB", 60000, "Intel i7")
]

# loop through all laptops
for lap in Laptops:
    print(lap.details())
    
# Filter (price < 60000)

for lap in Laptops:
    if lap.price < 60000:
        print(f"Laptops under 60000")
        print(lap.details())


#Insted of doing this:
#print(l1.details())
#print(l2.details())
#we can automate using loop
```
