## Create a python program using a class named lapto.

    taks :
        - define a class Laptop
        - Inside it, create a   method     details (self,    brand, ram)
        - create teo different laptop object.
        - Call the details() method for each object, passing different values (ex: "Dell", "16GB" and "Hp", "8GB")

```python
class Laptop:
    
    def details(self, brand, ram):
        self.brand = brand
        self.ram = ram
        print("Brand:", self.brand)
        print("RAM:", self.ram)
        print("------------------")

# Create objects
l1 = Laptop()
l2 = Laptop()

# Call method properly
l1.details("Dell", "16GB")
l2.details("HP", "8GB")
```
