## Create a class Greeter with a method greet(name) that prints a greeting for the provided name.

```python
class Greeter:
    def greet(self, name):
        print(f"Hello, {name}")
        
g = Greeter() # Object Creation

g.greet("Mayur") #Calling the method
```

### Return Value Instead of Print

```python
class Greeter:
    def greet(self, name):
        return f"Hello, {name}"
    
g = Greeter()  

name = input("Enter your name: ")

message = g.greet(name)

print(message)
```

### Real Use Case

```python
class Greeter:
    
    def greet(self, name):
        if name.strip() == "":
            return f"Hello, stranger! 😄"  # fixed : handele empty input
        
        name = name.capitalize()    # fixed: claen formatting
        
        if name == "Mayur":
            return f"Hello, {name}  Welcome back, developer"
        
        return f"Hello, {name}"
    
g = Greeter()

name = input("Enter your name:")
print(g.greet(name))
```

### Time-Based Greeting Code

```python
from datetime import datetime # New : import datetime

class Greeter:
    
    def greet(self, name):
        if name.strip() == "":
            name = "Stranger"
        else:
            name = name.capitalize()
            
        hour = datetime.now().hour  # get current time in hour
    
        if 5 <= hour < 12:
            greeting = "Good Morning"
        elif 12 <= hour < 17:
            greeting = "Good afternoon"
        elif 17<= hour < 21:
            greeting = "Good Evening"
        else:
            greeting = "Good Night"
        
        return f"{greeting}, {name}"

g = Greeter()

name = input("Enter your name")

print(g.greet(name))
        
```
