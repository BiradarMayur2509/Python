## Exercise : Develop a class Calculator with methods to add and subtract two numbers.

```python
class Calculator:
    
    def __init__(self):
        pass 
    
    def add(self, a, b):
        return a + b
    
    def subtract(self, a, b):
        return a - b
    
    def multiply(self, a,b):
        return a * b
    
# testing the calculator

calc = Calculator() # object creation

a = int(input("Enter first number: "))
b = int(input("Enter second number"))

print(f"Sum: " , calc.add(a,b))
print(f"sub : " , calc.subtract(a,b))
print(f"Multiiplication : " , calc.multiply(a,b))

```

## Menu-Based Calculator

```python
class Calculator:
    
    def add(self, a, b):
        return a + b
    
    def subtract(self, a, b):
        return a - b
    
    def multiply(self, a,b):
        return a * b
    
    def divide(self, a, b):
        if b == 0:
            return "Cannot divide by zero"
        return a / b
    
    
calc = Calculator()

while True:
    
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")
    
    choice = input("Enter your choice: ")
    
    if choice == "5":
        print("Exiting from calculator....")
        break
    
    a = int(input("Enter first number: "))
    b = int(input("Enter second number"))
    
    if choice == "1":
        print(f"Result: {cacl.add(a, b)}")
    elif choice == "2":
        print(f"Result: {calc.subtract(a,b)}")
    elif choice == "3":
        print(f"Result: {calc.multiply(a,b)}")
    elif choice == "4":
        print(f"Result: {calc.divide(a,b)}")
    else:
        print("Invalid choice! Try again")
```
